```python
import pandas as pd
import duckdb
```

```python
conn = duckdb.connect("../data/warehouse/retail.duckdb")
```

```python
print(conn.execute("""
    SELECT table_name
    FROM information_schema.tables
    WHERE table_schema = 'bronze'
    ORDER BY table_name
""").df())
```

# 1. Orders 

```python
orders_df = conn.execute(
    """
    SELECT * FROM bronze.orders LIMIT 5
    """
).df()

orders_df.head()
```

```python
orders_df['order_status'].unique(),\
orders_df['payment_method'].unique(),\
orders_df['device_type'].unique(),\
orders_df['order_source'].unique()
```
